#### 1. Create a null vector of size 10 but the fifth value which is 1 (base)

#### 2. Create a 3x3 matrix with values ranging from 0 to 8 with zeros on the diagonal (base)
`hint: np.arange, reshape, np.eye`

```python
M = np.reshape(np.arange(0,9),(3,3))
M = M - M*np.eye(3,3)
M
```

#### 3. How to add a border (filled with 0's) around an 5x5 matrix of ones? (base)
`hint: np.ones, np.pad`

```python
Z = np.ones((5,5))
Z_padded = np.pad(Z,1,'constant',constant_values=0)

Z_padded
```

#### 4. Normalize a 5x5 random matrix (base)
`hint: (x - mean) / std, np.mean, np.std`

```python
R = np.random.rand(5,5)
mean = np.mean(R)
std = np.std(R)
R_norm = (R - mean)/std

R_norm
```

#### 5. Multiply a 5x3 matrix of ones by a 3x2 matrix of ones (real matrix product) (base)
`hint: np.dot`

```python
A = np.ones((5,3))
B = np.ones((3,2))
P = np.dot(A,B)

P
```

#### 8. Create random vector of size 10 and replace the maximum value by its additive inverse and the minimum value (of the original array) with the median value (intermediate)
`hint: argmax, argmin`

```python
randomVector = np.random.rand(10)
max_idx = np.argmax(randomVector)
min_idx = np.argmin(randomVector)
median = np.median(randomVector)

print(f"Original vector: {randomVector}")
print(f"max idx = {max_idx}, min idx = {min_idx}")

randomVector[max_idx] = -randomVector[max_idx]
randomVector[min_idx] = median

print(f"\nModified vector: {randomVector}")
```

#### 9. Randomly replace p elements in a 2D nxn zero matrix to 1 (intermediate)
`hint: np.put, np.random.choice`

```python
n = int(input("Insert n: "))
p = int(input("Insert p: "))

assert p <= n**2

zeroMatrix = np.zeros((n,n))

idxs = np.random.choice(range(0,n**2),p,replace=False)

np.put(zeroMatrix,idxs,np.ones(p))

zeroMatrix
```

#### 10. Subtract the mean of each row of a random 5x10 matrix (intermediate)
`hint: mean(axis=,keepdims=)`

```python
randomMatrix = np.random.rand(5,10)
print("*- Original matrix -*")
print(randomMatrix)

means = np.mean(randomMatrix,axis=1,keepdims=True)
print("\n*- Row means -*")
print(np.reshape(means,(-1,1)))

randomMatrix_sub = randomMatrix - means
print("\n*- Modified matrix -*")
print(randomMatrix_sub)
```
