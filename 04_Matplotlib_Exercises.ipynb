## Base

### Plot a line with formula $y = 2x + 1$
`hint: np.arange, plt.figure, plt.plot, plt.title, plt.xlabel, plt.ylabel, plt.grid, plt.legend, plt.show`

- Use points in range [1, 20]
- Figure size of (10,5)
- Set axis labels
- Set plot title
- Set line color as red
- Discontinued line (--) with star (\*) on point
- Plot the legend with the formula in latex version (`r"$ ... $"`)
- Set a dashed grid

```python
x = np.arange(1,21)
y1 = 2*x + 1
```

```python
fig = plt.figure(1,figsize=(10,5))
plt.xlabel(r"$x$"); plt.ylabel(r'$y$')
plt.title('First plot');
plt.plot(x,y1,'r*--',label=r"$y=2x+1$")
plt.legend()
plt.grid(linestyle='dashed')
plt.show()
```

### Add to the previous plot the line with formula $y = log(x) + 1$ in the same range
`hint: np.log, plt.plot`

- Set line color as blue
- Normal line (-) with square (s) char on point

```python
y2 = np.log(x) + 1

plt.figure(fig)
plt.plot(x,y2, linestyle = '-', marker="s", color='b',label=r"$y=log(x)+1$")
plt.legend()
plt.show()
```

### Add to the previous plot the line with formula $y=x^2 - 1$ in the same range
`hint: x**2, plt.plot`

- Set line color as green
- Normal line (-) with + (plus) char on point

```python
y3 = x**2 - 1

plt.figure(fig)
plt.plot(x, y3, linestyle = '-',color='g', marker='+',label=r"$x^2 -1$")
plt.legend()
plt.show()
```

### Replot the previous plot in different subplots in the same line without sharing axes
`hint: plt.subplots, axes[].plot, axes[].set_title, axes[].set_xlabel, axes[].set_ylabel, axes[].grid, axes[].legend, fig.tight_layout`

```python
fig, axes = plt.subplots(1, 3, figsize=(12, 4))

axes[0].plot(x, y1, label=r"$2x+1$")
axes[0].set_title("Linear function")
axes[0].set_xlabel(r"$x$"); axes[0].set_ylabel(r'$y$')
axes[0].grid(linestyle='dashed')
axes[0].legend()

axes[1].plot(x, y2, label=r"$log(x)+1$")
axes[1].set_title("Logarithmic function")
axes[1].set_xlabel(r"$x$"); axes[1].set_ylabel(r'$y$')
axes[1].grid(linestyle='dashed')
axes[1].legend()

axes[2].plot(x, y3, label=r"$x^2 -1$")
axes[2].set_title("Quadratic function")
axes[2].set_xlabel(r"$x$"); axes[2].set_ylabel(r'$y$')
axes[2].grid(linestyle='dashed')
axes[2].legend()

fig.tight_layout()
plt.show()
```

### Replot the previous plot sharing y between suplots
`hint: sharey=True`

```python
fig, axes = plt.subplots(1, 3, figsize=(12, 4), sharey=True)

axes[0].plot(x, y1, label=r"$2x+1$")
axes[0].set_title("Linear function")
axes[0].set_xlabel(r"$x$"); axes[0].set_ylabel(r'$y$')
axes[0].grid(linestyle='dashed')
axes[0].legend()

axes[1].plot(x, y2, label=r"$log(x)+1$")
axes[1].set_title("Logarithmic function")
axes[1].set_xlabel(r"$x$"); axes[1].set_ylabel(r'$y$')
axes[1].grid(linestyle='dashed')
axes[1].legend()

axes[2].plot(x, y3, label=r"$x^2 -1$")
axes[2].set_title("Quadratic function")
axes[2].set_xlabel(r"$x$"); axes[2].set_ylabel(r'$y$')
axes[2].grid(linestyle='dashed')
axes[2].legend()

fig.tight_layout()
plt.show()
```

### Replot the previous plot using the for loop, lists and dictionaries
`hint: functions = [...], {"title": ..., "y": ..., "label": ..., "linestyle": ...}, zip(functions, axes)`

```python
functions = [
    {"title": "Linear function", "y": y1, "label": r"$2x+1$", "linestyle": '-'},
    {"title": "Logarithmic function", "y": y2, "label": r"$log(x)+1$", "linestyle": '-'},
    {"title": "Quadratic function", "y": y3, "label": r"$x^2 -1$", "linestyle": '-'}
]

fig, axes = plt.subplots(1, 3, figsize=(12, 4), sharey=True)
graphs = zip(functions,axes)

for func, ax in graphs:
    ax.plot(x, func["y"], label=func["label"], linestyle=func["linestyle"])
    ax.set_title(func["title"])
    ax.set_xlabel(r"$x$"); ax.set_ylabel(r'$y$')
    ax.grid(linestyle='dashed')
    ax.legend()

fig.tight_layout()
```

### Plot functions $ y_i = i\cdot x - 5i$ with $i$ in range (0, 10) in the same plot with size (10,5) and a different color for each function

```python
fig, ax = plt.subplots(figsize=(10,5))
for i in range(10):
    ax.plot(x, i*x - 5*i, label = rf"$y_{i} = {i}x -{5*i}$")

ax.legend()
```

### Plot functions $ y_i = i\cdot x - 5i$ with $i$ in range (0, 10) in different subplots with shared y and figure size (20,5)

```python
fig, axes = plt.subplots(1,10,figsize=(20,5), sharey=True)
for i in range(10):
    axes[i].plot(x, i*x - 5*i, color = f"C{i}")
    axes[i].set_xlabel("$x$")
    axes[i].set_ylabel("$y$")
    axes[i].set_title(f"$y_{i} = {i}x - {5*i}$")
    axes[i].grid()


fig.tight_layout()
plt.show()
```

## Intermediate

### Plot a red scatterplot of $x$ squared
`hint: plt.scatter`

```python
plt.figure(figsize=(7.5,5))
plt.xlabel(r"$x$"); plt.ylabel(r'$y$')
plt.scatter(x, x**2, color='r',label=r"$y[x]=x^2$")
plt.title(r"Scatterplot of $x^2$")
plt.grid()
plt.legend()
```

### Plot a blue step plot of $x$ squared with linewidth of $3$
`hint: plt.step`

```python
plt.figure(figsize=(7.5,5))
plt.xlabel(r"$x$"); plt.ylabel(r'$y$')
plt.step(x, x**2, color='b',label=r"$y_{\epsilon}[x]=x^2$", linewidth=3)
plt.title(r"Step plot of $x^2$")
plt.grid()
plt.legend()
```

### Plot an orange barplot plot of $x$ squared with alpha of $0.25$
`hint: plt.bar`

```python
plt.figure(figsize=(7.5,5))
plt.xlabel(r"$x$"); plt.ylabel(r'$y$')
plt.bar(x, x**2, color='orange',label=r"$y_{bar}[x]=x^2$", alpha=0.25)
plt.title(r"Bar plot of $x^2$")
plt.grid()
plt.legend()
```

### Plot the area between $x$ squared and the $x \cdot log(x)$ function in green and alpha $0.5$
`hint: plt.fill_between`

```python
plt.figure(figsize=(7.5,5))
plt.xlabel(r"$x$"); plt.ylabel(r'$y$')
plt.fill_between(x, x**2, x*np.log(x), color='g', alpha=0.5)
plt.title(r"Area between $x^2$ and $x log(x)$")
plt.grid()
```

### Replot previous plots in square grid (2,2)
`hint: axes[][]`

```python
fig, axes = plt.subplots(2,2, figsize=(10,10))

axes[0][0].scatter(x, x**2, color='r',label=r"$y[x]=x^2$")
axes[0][0].set_xlabel(r"$x$"); axes[0][0].set_ylabel(r'$y$')
axes[0][0].set_title(r"Scatterplot of $x^2$")
axes[0][0].grid()
axes[0][0].legend()

axes[0][1].step(x, x**2, color='b',label=r"$y_{\epsilon}[x]=x^2$", linewidth=3)
axes[0][1].set_xlabel(r"$x$"); axes[0][1].set_ylabel(r'$y$')
axes[0][1].set_title(r"Step plot of $x^2$")
axes[0][1].grid()
axes[0][1].legend()

axes[1][0].bar(x, x**2, color='orange',label=r"$y_{bar}[x]=x^2$", alpha=0.25)
axes[1][0].set_xlabel(r"$x$"); axes[1][0].set_ylabel(r'$y$')
axes[1][0].set_title(r"Bar plot of $x^2$")
axes[1][0].grid()
axes[1][0].legend()

axes[1][1].fill_between(x, x**2, x*np.log(x), color='g', alpha=0.5)
axes[1][1].set_xlabel(r"$x$"); axes[1][1].set_ylabel(r'$y$')
axes[1][1].set_title(r"Area between $x^2$ and $x log(x)$")
axes[1][1].grid()


fig.tight_layout()
```

### Plot purple histogram of 50 bins and pink cumulative instogram of 100K random samples in subplots
`hint: np.random.randn, axes[].hist(), cumulative=True, bins=50`

```python
samples = np.random.randn(100000)

fig, axes = plt.subplots(1,2, figsize=(10,5))

axes[0].hist(samples, color='purple', bins=50)
axes[0].set_xlabel("x"); axes[0].set_ylabel("frequency")
axes[0].set_title("Histogram of the samples")
axes[0].grid()

axes[1].hist(samples, color='pink', bins=50 ,cumulative=True)
axes[1].set_xlabel("x"); axes[1].set_ylabel("cumulative frequency")
axes[1].set_title("Cumulative histogram")
axes[1].grid()

fig.tight_layout()
```
